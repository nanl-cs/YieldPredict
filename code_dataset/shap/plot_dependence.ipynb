```python
import pandas as pd
import shap
import matplotlib.pyplot as plt
import os

# ==========================================
# 配置接口：在此处修改你需要分析的特征名称
# ==========================================
# TARGET_FEATURE = 'Cum_VPD_P4'  # 例如：'P4_PPT', 'P1_GDD', 'P3_VPD' 等
# TARGET_FEATURE = 'P1_GDD'  # 例如：'P4_PPT', 'P1_GDD', 'P3_VPD' 等
# TARGET_FEATURE = 'P4_WUE'  # 例如：'P4_PPT', 'P1_GDD', 'P3_VPD' 等
TARGET_FEATURE = 'P4_NDWI'  # 例如：'P4_PPT', 'P1_GDD', 'P3_VPD' 等
# ==========================================

def plot_shap_dependence(target_feature):
    # 1. 检查并读取底层矩阵数据
    if not os.path.exists("SHAP_Values_Matrix.csv") or not os.path.exists("SHAP_Feature_Matrix.csv"):
        print("错误: 找不到矩阵文件，请确认当前目录存在 SHAP_Values_Matrix.csv 和 SHAP_Feature_Matrix.csv")
        return
        
    shap_df = pd.read_csv("SHAP_Values_Matrix.csv")
    feat_df = pd.read_csv("SHAP_Feature_Matrix.csv")
    
    if target_feature not in feat_df.columns:
        print(f"错误: 特征 '{target_feature}' 不存在。请从以下列表中选择: \n{list(feat_df.columns)}")
        return

    # 2. 转换数据格式以适配 SHAP 库
    shap_values = shap_df.values
    
    # 3. 绘制偏依赖图
    # 设置一个较大的画布以容纳图例
    fig, ax = plt.subplots(figsize=(8, 6))
    
    # dependence_plot 会自动计算并选择一个交互作用最强的特征作为颜色映射 (右侧 Y 轴)
    shap.dependence_plot(
        target_feature, 
        shap_values, 
        feat_df, 
        ax=ax,
        show=False,
        alpha=0.7,      # 透明度，防止点过度重叠
        dot_size=15     # 点的大小
    )
    
    # 4. 样式调整与保存
    plt.title(f"SHAP Dependence Plot: {target_feature}", fontsize=14, pad=15)
    plt.grid(True, linestyle='--', alpha=0.3)
    plt.tight_layout()
    
    output_file = f"SHAP_Dependence_{target_feature}.pdf"
    plt.savefig(output_file, dpi=300, bbox_inches='tight')
    plt.close()
    
    print(f"偏依赖图已成功生成并保存至: {output_file}")

if __name__ == "__main__":
    plot_shap_dependence(TARGET_FEATURE)
```
